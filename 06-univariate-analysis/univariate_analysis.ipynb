# Univariate Analysis

## Objective

Learn how to analyze individual numerical and categorical variables.

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv("../datasets/sample_data.csv")
```

```python
# Calculate the average age
df["Age"].mean()
```

```python
# Find the median age
df["Age"].median()
```

```python
# Find the most frequent age
df["Age"].mode()
```

```python
# Find minimum and maximum age
df["Age"].min(), df["Age"].max()
```

```python
# Calculate standard deviation
df["Age"].std()
```

```python
# Calculate variance
df["Age"].var()
```

```python
# Display frequency of each department
df["Department"].value_counts()
```

```python
# Plot histogram for Age
plt.figure(figsize=(8,5))
plt.hist(df["Age"], bins=10)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# Plot boxplot for Salary
plt.figure(figsize=(8,5))
sns.boxplot(x=df["Salary"])
plt.title("Salary Box Plot")
plt.show()
```

```python
# Plot count plot for Department
plt.figure(figsize=(8,5))
sns.countplot(x="Department", data=df)
plt.title("Department Count")
plt.xticks(rotation=45)
plt.show()
```

# Observation

- Calculated summary statistics.
- Analyzed numerical variables.
- Analyzed categorical variables.
- Visualized distributions using Histogram, Box Plot, and Count Plot.

# Conclusion

Univariate Analysis helps understand the characteristics of individual variables and prepares the dataset for relationship analysis.
