# What is EDA?

Exploratory Data Analysis (EDA) is the process of understanding, exploring, cleaning, and summarizing a dataset before applying machine learning algorithms.

# Why Do We Perform EDA?

- Understand the dataset
- Find missing values
- Remove duplicates
- Detect outliers
- Verify data types
- Generate useful insights
