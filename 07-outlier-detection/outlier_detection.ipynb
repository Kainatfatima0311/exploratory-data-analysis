# Outlier Detection

## Objective

Learn how to identify and handle outliers using visualization and statistical methods.

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
df = pd.read_csv("../datasets/sample_data.csv")
```

```python
# Display a box plot for the Salary column
plt.figure(figsize=(8,5))
sns.boxplot(x=df["Salary"])
plt.title("Salary Box Plot")
plt.show()
```

```python
# Calculate the first and third quartiles
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

# Calculate the Interquartile Range
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
# Calculate lower and upper limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
# Display outliers
outliers = df[
    (df["Salary"] < lower_limit) |
    (df["Salary"] > upper_limit)
]

outliers
```

```python
# Remove outliers
clean_df = df[
    (df["Salary"] >= lower_limit) &
    (df["Salary"] <= upper_limit)
]

clean_df
```

```python
# Compare the number of rows before and after removing outliers
print("Original Rows:", len(df))
print("Rows After Removing Outliers:", len(clean_df))
```

# Observation

- Box plot displayed potential outliers.
- IQR method calculated acceptable limits.
- Outliers were identified.
- Clean dataset created after removing outliers.

# Conclusion

Outliers were successfully detected using the Box Plot and IQR methods. The cleaned dataset is now more suitable for further analysis.
