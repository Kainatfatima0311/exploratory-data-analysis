# Duplicate Data

## Objective

Learn how to detect, inspect, and remove duplicate records using Pandas.

```python
# Import Pandas
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv("../datasets/sample_data.csv")
```

## Check Duplicate Rows

```python
# Check duplicate rows
df.duplicated()
```

```python
# Count duplicate rows
df.duplicated().sum()
```

```python
# Display duplicate rows
df[df.duplicated()]
```

```python
# Remove duplicate rows
df = df.drop_duplicates()
```

```python
# Verify duplicate rows after removal
df.duplicated().sum()
```

# Observation

- Duplicate rows identified.
- Total duplicate records counted.
- Duplicate rows displayed.
- Duplicate rows removed successfully.

# Conclusion

Duplicate records were successfully identified and removed from the dataset.
