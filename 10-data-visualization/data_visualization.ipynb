# Data Visualization

## Objective

Learn how to create professional charts using Matplotlib and Seaborn.

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv("../datasets/sample_data.csv")
```

```python
# Line Plot
plt.figure(figsize=(8,5))

plt.plot(df["Experience_Years"], df["Salary"])

plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")

plt.tight_layout()

plt.show()
```

```python
# Bar Plot
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="Department", y="Salary")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
# Count Plot
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="Department")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
# Histogram
plt.figure(figsize=(8,5))

plt.hist(df["Age"], bins=10)

plt.title("Age Distribution")

plt.show()
```

```python
# Box Plot
plt.figure(figsize=(8,5))

sns.boxplot(x=df["Salary"])

plt.show()
```

```python
# Violin Plot
plt.figure(figsize=(8,5))

sns.violinplot(
    data=df,
    x="Department",
    y="Salary"
)

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
# Scatter Plot
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Experience_Years",
    y="Salary"
)

plt.show()
```

```python
# Heatmap
plt.figure(figsize=(8,6))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.show()
```

```python
# Pie Chart
department_counts = df["Department"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    department_counts,
    labels=department_counts.index,
    autopct="%1.1f%%"
)

plt.show()
```

# Observation

- Compared numerical and categorical data.
- Visualized distributions and relationships.
