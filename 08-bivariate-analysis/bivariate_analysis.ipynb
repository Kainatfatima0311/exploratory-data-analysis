# Bivariate Analysis

## Objective

Learn how to analyze relationships between two variables using Pandas and Seaborn.

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv("../datasets/sample_data.csv")
```

```python
# Display scatter plot for Experience vs Salary
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Experience_Years", y="Salary")
plt.title("Experience vs Salary")
plt.show()
```

```python
# Display average salary by department
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Department", y="Salary")
plt.title("Average Salary by Department")
plt.xticks(rotation=45)
plt.show()
```

```python
# Display salary distribution by department
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Department", y="Salary")
plt.title("Salary Distribution by Department")
plt.xticks(rotation=45)
plt.show()
```

```python
# Calculate correlation between Age and Salary
df[["Age", "Salary"]].corr()
```

```python
# Calculate average salary by department
df.groupby("Department")["Salary"].mean()
```

```python
# Display frequency table for Gender and Department
pd.crosstab(df["Gender"], df["Department"])
```

# Observation

- Examined relationships between numerical variables.
- Compared average salary across departments.
- Measured correlation between Age and Salary.
- Analyzed categorical relationships using Crosstab.
