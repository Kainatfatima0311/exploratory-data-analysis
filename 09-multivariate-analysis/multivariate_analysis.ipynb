# Multivariate Analysis

## Objective

Learn how to analyze relationships among multiple variables using Pandas and Seaborn.

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv("../datasets/sample_data.csv")
```

```python
# Display correlation matrix
df.corr(numeric_only=True)
```

```python
# Visualize correlation matrix using heatmap
plt.figure(figsize=(10,6))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
# Pairplot for selected numerical features
sns.pairplot(
    df[
        [
            "Age",
            "Salary",
            "Experience_Years",
            "Performance_Score"
        ]
    ]
)

plt.show()
```

```python
# Scatter plot with Department as hue
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Experience_Years",
    y="Salary",
    hue="Department"
)

plt.title("Experience vs Salary by Department")

plt.show()
```

```python
# Average salary by Department and Gender
df.groupby(
    ["Department","Gender"]
)["Salary"].mean()
```

```python
# Create pivot table
pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    columns="Gender",
    aggfunc="mean"
)
```

# Observation

- Generated a correlation matrix.
- Visualized relationships using a heatmap.
- Compared multiple numerical variables using a pairplot.
- Analyzed salary by department and gender.
- Created a pivot table for summarized insights.
