# Missing Values

## Objective

Learn how to detect and handle missing values using Pandas.

```python
#libraries
import pandas as pd 
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
# Load the dataset
df = pd.read_csv("../datasets/sample_data.csv")
```

## Display Missing Values

```python
# Check missing values in every cell
df.isnull()
```

```python
# Count missing values in each column
df.isnull().sum()
```

```python
# Calculate missing value percentage
(df.isnull().sum() / len(df)) * 100
```

## Fill Numerical Missing Values

```python
# Fill missing values in Age column using mean
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
# Fill missing values in Salary column using median
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
```

## Fill Categorical Missing Values

```python
# Fill missing values in City column using mode
df["City"] = df["City"].fillna(df["City"].mode()[0])
```

```python
# Verify that missing values have been handled
df.isnull().sum()
```

# Observation

- Missing values detected.
- Missing values counted.
- Numerical columns filled.
- Categorical columns filled.
- Dataset verified.

# Conclusion

Missing values were successfully detected and handled using appropriate techniques.
