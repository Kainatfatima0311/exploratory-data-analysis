# Data Cleaning

## Objective

Learn how to clean and standardize a dataset using Pandas.

```python
# Import library
import pandas as pd
```

```python
#load dataset
df = pd.read_csv("../datasets/sample_data.csv")
```

```python
# Display unique city names before cleaning
df["City"].unique()
```

```python
# Remove extra spaces
df["City"] = df["City"].str.strip()
```

```python
# Convert all city names to title case
df["City"] = df["City"].str.title()
```

```python
# Replace incorrect spellings
df["City"] = df["City"].replace({
    "Lhore": "Lahore",
    "Karach": "Karachi",
    "Islambad": "Islamabad"
})
```

```python
# Display cleaned city names
df["City"].unique()
```

```python
# Count records in each department
df["Department"].value_counts()
```

```python
# Convert Joining_Date to datetime
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])
```

```python
# Check updated data types
df.dtypes
```

# Observation

- Removed extra spaces.
- Standardized city names.
- Fixed incorrect spellings.
- Converted date column.
- Verified updated data types.

# Conclusion

The dataset has been cleaned and standardized, making it ready for further analysis.
